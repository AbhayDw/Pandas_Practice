```python
def strength(n):
  if n>150:
    return "Strong"
  elif n>100:
    return "Average"
  else:
    return "Weak"
```

```python
df["Power"]=df["Attack"].apply(strength)
```

```python
df
```

```python
def defensive(n):
  if n>120:
    return "Defensive"
  elif 80<=n<120:
    return "Avg Defensive"
  else:
    return "less Defensive"
```

```python
df["Defensive_pokemon"]=df["Defense"].apply(defensive)
```

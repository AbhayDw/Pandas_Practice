### **sorting using sort_values functions**

```python
s2=s1.sort_values(axis=0)#sort name wise
s2
```

```python
s2=s1.sort_values(axis=0,ascending=False)
s2
```

```python
#we also use kind foor specific sorting like quick merge
s2=s1.sort_values(axis=0,ascending=False,kind="quicksort")
s2
```

```python
##meaning of in place
arr=[1,6,8,3,None,4,None,2,9]
s3=pd.Series(arr)
s3
```

```python
#sabhi nan value ok last me karne ke liye temprory
s3.sort_values(axis=0,na_position="last")# not in_place
```

```python
##in_place
s4=pd.Series(arr)
s4
```

```python
s4.sort_values(axis=0,na_position="last",inplace=True)
s4
```

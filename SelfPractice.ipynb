### Fetching a single column
You can access a single column by using square brackets with the column name. This will return a pandas Series.

```python
display(df["name"]) # Fetching the 'name' column
```

### Fetching multiple columns
To fetch multiple columns, you need to pass a list of column names inside the square brackets. This will return a new DataFrame containing only those columns.

```python
display(df[["name", "age", "sex"]]) # Fetching 'name', 'age', and 'sex' columns
```

```python
# but i loc you need to write collumn name
df.loc[1:10,("name","sex")]
```

### **filtering**

```python
df=pd.read_excel("/content/titanic.xls",usecols=["pclass","survived","name","fare","sex","age"])
df
```

```python

df
```

```python
df.iloc[2]
```

```python
#print all the males
#using boolean variable
males_df=df[df["sex"]=="male"]
males_df
```

```python
females_df=df[df["sex"]=="female"]
females_df
```

```python
rich_kids_df=df[(df["age"]<20) & (df["fare"]>200)]
rich_kids_df
```

```python
poor_kids_df=df[(df["age"]<20)&(df["fare"]<20)]
poor_kids_df
```

```python
old_age_df=df[(df["age"]>60) & (df["fare"]>200)]
old_age_df
```

```python
old_age_df=df[(df["age"]>60) & (df["pclass"]==3)]
old_age_df
```

```python
rich_p=df[(df["pclass"]==1)]
rich_p
```
